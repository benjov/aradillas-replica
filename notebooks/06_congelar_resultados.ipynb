Corre las configuraciones de `codigo/congelar_resultados.py` y compara **cada valor** con el congelado versionado en `resultados/`. La corrida nueva se escribe en `resultados/nueva_corrida/` para no pisar la referencia.

| qué se corre | tiempo (Mac, Python 3.9.6) | necesita |
|---|---|---|
| solo 2014 (4 configuraciones) | ~2.5 min | `datos/Data_2014/` |
| todas (7 configuraciones) | ~15 min | además `datos/Data_2022/` |

**Cómo leer el resultado.** Con el mismo entorno del congelado (ver notebook 00) se esperan **0 diferencias**. Con otro sistema operativo o versión de numpy pueden aparecer diferencias de redondeo en los últimos decimales: sube `TOLERANCIA` para distinguirlas de un cambio real y **no las trates como reproducción exacta**.

Solo si el código cambió a propósito y la corrida es la buena, se adopta como nueva referencia: `cp resultados/nueva_corrida/* resultados/`, y se hace commit antes de citar cifras.

```python
# Preparación (no modificar): ubica la raíz del repo, la usa como directorio de trabajo
# y pone codigo/ en el path. Funciona igual abriendo el notebook desde notebooks/ o desde
# la raíz, y también ejecutándolo con codigo/run_nb.py.
import os, sys
_raiz = os.getcwd()
while not os.path.isdir(os.path.join(_raiz, "codigo")) and os.path.dirname(_raiz) != _raiz:
    _raiz = os.path.dirname(_raiz)
os.chdir(_raiz)
sys.path.insert(0, os.path.join(_raiz, "codigo"))
print("Raíz del repo:", _raiz)
```

```python
# Configuración de la corrida
CONFIGS = None          # None = todas; o una lista, p. ej. ["replica_2014", "comparable_2014"]
SOLO_2014 = False       # True: solo las cuatro configuraciones de 2014 (no requiere Data_2022)
TOLERANCIA = 0.0        # 0.0 = comparación exacta (la del congelado canónico)
```

```python
import os, sys, time
os.environ["ARADILLAS_SALIDA"] = "resultados/nueva_corrida"     # antes de importar el script
import congelar_resultados as cr
import verificar

nombres = CONFIGS or list(cr.CONFIGS)
if SOLO_2014:
    nombres = [n for n in nombres if cr.CONFIGS[n]["anio"] == 2014]
print("Configuraciones a correr:", nombres)
t0 = time.time()
cr.main(nombres)
print(f"\nTiempo total: {(time.time() - t0) / 60:.1f} min")
```

```python
res = verificar.comparar("resultados/resultados_congelados.json",
                         "resultados/nueva_corrida/resultados_congelados.json",
                         configs=nombres, tol=TOLERANCIA)
verificar.informe(res)
```
